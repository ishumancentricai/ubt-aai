### What is Overfitting/Underfitting?
As mentioned, in statistics and machine learning we usually split our data into subsets (training data and testing data) and fit our model on the train data, in order to make predictions on the test data. When we do that, one of two things might happen: we overfit our model or we underfit our model. We don’t want any of these things to happen, because they affect the predictability of our model — we might be using a model that has lower accuracy and/or is ungeneralized (meaning you can’t generalize your predictions on other data). Let’s see what under and overfitting actually mean:

#### Overfitting
Overfitting means that model we trained has trained “too well” and is now fit too closely to the training dataset. This usually happens when the model is too complex (i.e., too many features/variables compared to the number of observations). This model will be very accurate on the training data but will probably be not accurate on untrained or new data. It is because this model is not generalized (or not AS generalized), meaning you can generalize the results and can’t make any inferences on other data, which is, ultimately, what you are trying to do. Basically, when this happens, the model learns or describes the “noise” in the training data instead of the actual relationships between variables in the data. This noise, obviously, isn’t part in of any new dataset, and cannot be applied to it.

#### Underfitting
In contrast to overfitting, when a model is underfitted, it means that the model does not fit the training data and therefore misses the trends in the data. It also means the model cannot be generalized to new data. As you probably guessed (or figured out!), this is usually the result of a very simple model (not enough predictors/independent variables). It could also happen when, for example, we fit a linear model (like linear regression) to data that is not linear. It almost goes without saying that this model will have poor predictive ability (on training data and can’t be generalized to other data).

![](https://cdn-images-1.medium.com/max/800/1*tBErXYVvTw2jSUYK7thU2A.png)

It is worth noting the underfitting is not as prevalent as overfitting. Nevertheless, we want to avoid both of those problems in data analysis. You might say we are trying to find the middle ground between under and overfitting our model. As you will see, train/test split and cross validation help to avoid overfitting more than underfitting. Let’s dive into both of them!

#### Training, Validation, and Testing
From development to deployment of ML models, we usually need data for three procedures:
- **Training**: Used to fit or train the model’s parameters.
- **Validation**: Used to tune model hyperparameters and select the best model configuration---not used for training during model optimization!
- **Testing**: Used only at the very end to evaluate the final model’s performance on **unseen** data (test data)---Never use test set data during training or validation!

### Holdout Sets: Train-Test Split
The easiest way to split data is by generating holdout sets, where one part of the data is used for training and the rest can be used for testing (or validation):

![](https://miro.medium.com/v2/1*NQaN71ejH_eTUxhRLwiJcA.png)

A standard procedure would look like this:
1. Split the entire dataset into two parts: train & test set (keep the test set untouched until the very end!)
2. Split the train set once again into two parts: train & validation set
3. Fit the model on the train set and find optimal parameters with the validation set
4. Evaluate final performane on test set

Let’s see how to do this in Python. We’ll do this using the Scikit-Learn library and specifically the train_test_split method. We’ll start with importing the necessary libraries:

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
````

````python
# 1. Apply a train test split with a test size of 0.2
df = pd.read_csv(data_path, index_col=0)
df.head(5)
````

````python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df['target'], test_size=0.2)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2)
````

````python
# 2. Print shapes of resulting data sets
print("Shape of X_train:", X_train.shape)
print("Shape of X_val:", X_val.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_val:", y_val.shape)
print("Shape of y_test:", y_test.shape)
````

````python
# 3. Fit a linear regression model and a MLPRegressor
from sklearn.linear_model import LinearRegression

# Fit a linear regression model
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
````

````python
from sklearn.neural_network import MLPRegressor

# Fit a MLPRegressor model
mlp_model = MLPRegressor(max_iter=1000)
mlp_model.fit(X_train, y_train)
````

````python
# Compare the MAE of the two models
from sklearn.metrics import mean_absolute_error

# Predict the target variable using the linear regression model
linear_predictions = linear_model.predict(X_test)

# Predict the target variable using the MLPRegressor model
mlp_predictions = mlp_model.predict(X_test)

# Calculate the mean absolute error for each model
linear_mae = mean_absolute_error(y_test, linear_predictions)
mlp_mae = mean_absolute_error(y_test, mlp_predictions)

# Compare the MAE of the two models
if linear_mae < mlp_mae:
  print("Linear Regression model has a lower MAE:", linear_mae)
else:
  print("MLPRegressor model has a lower MAE:", mlp_mae)
````

Lower MAE indicates that a model makes smaller prediction errors on average.

**Problem:** We fit the model and immediately tested it, but without a validation set we can't tune hyperparameters or select the best model configuration.

**Why use validation?** The validation set lets us compare different models or parameters and choose the best one before touching the test set. We don't use the test set to do this, to prevent overfitting to the test data and to get an honest final performance estimate.

````python
# Lets find an "optimal" architecture for MLPRegressor using the validation set, based on MAE
hidden_layers = [(100,), (50, 50)]
alpha = [0.0001, 0.001]
best_score = float('inf')
best_combination = None

for layers in hidden_layers:
  for a in alpha:
    model = MLPRegressor(hidden_layer_sizes=layers, alpha=a, max_iter=1000, random_state=42)
    model.fit(X_train, y_train)

    val_predictions = model.predict(X_val)
    val_mae = mean_absolute_error(y_val, val_predictions)
    print(f"hidden_layers={layers}, alpha={a}: Validation MAE = {val_mae:.4f}")

  if val_mae < best_score:
    best_score = val_mae
    best_combination = (layers, a)

print(f"\nBest configuration: hidden_layer_sizes={best_combination[0]}, alpha={best_combination[1]}")

# Train final model
mlp_model = MLPRegressor(hidden_layer_sizes=best_combination[0], alpha=best_combination[1], max_iter=2000, random_state=42)
mlp_model.fit(X_train, y_train)

final_predictions = mlp_model.predict(X_test)
final_mae = mean_absolute_error(y_test, final_predictions)
print(f"Test MAE = {final_mae:.4f}")
````

### Cross Validation

Instead of simply splitting data into two subsets, we can also perform something called cross validation. It’s very similar to train/test split, but it’s applied to more subsets. Meaning, we split our data into k subsets, and train on k-1 one of those subset. What we do is to hold the last subset for test. We’re able to do it for each of the subsets.

#### K-Folds Cross Validation
In K-Folds Cross Validation we split our data into k different subsets (or folds). We use k-1 subsets to train our data and leave the last subset (or the last fold) as test data. We then average the model against each of the folds and then finalize our model. After that we test it against the test set.

![](https://cdn-images-1.medium.com/max/800/1*J2B_bcbd1-s1kpWOu_FZrg.png)

Here is a very simple example from the Sklearn documentation for K-Folds:

````python
from sklearn.model_selection import KFold 
import numpy as np

X = np.array([[1, 2], [3, 4], [1, 2], [3, 4]])
y = np.array([1, 2, 3, 4])
kf = KFold(n_splits=2)
kf
````

And let’s see the result — the folds:

````python
for train_index, test_index in kf.split(X):
  print("TRAIN:", train_index, "TEST:", test_index)
  X_train, X_test = X[train_index], X[test_index]
  y_train, y_test = y[train_index], y[test_index]
````

As you can see, the function split the original data into different subsets of the data. Again, very simple example but I think it explains the concept pretty well.

### **TASK: Cross Validation**

1. Apply a 6-fold cross validation
2. Fit a [random forest regressor](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html) model and calculate the MAE for each fold and the average MAE score

````python
# 1. Apply a 6-fold cross validation
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold 

kf = KFold(n_splits=6)
````

````python
# 2. Fit a random forest regressor model and calculate the MAE for each fold and the average MAE score
model = RandomForestRegressor()
rf_mae_list = []

df = pd.read_csv(data_path, index_col=0)
X = df.iloc[:, :-1]
y = df['target']


for train_index, test_index in kf.split(X):
  X_train, X_test = X.iloc[train_index], X.iloc[test_index]
  y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
  # Fit a random forrest regressor
  model.fit(X_train, y_train)

  y_pred = model.predict(X_test)
    
  rf_mae = mean_absolute_error(y_test, y_pred)
  print("MAE:", rf_mae)
    
  rf_mae_list.append(rf_mae)

# Print the scores
print(f"Cross Validation average mae score:", np.mean(rf_mae_list))
````

3.
Cross-validation splits the data into multiple folds and trains/validates on different combinations, giving a more robust performance estimate. Unlike a single holdout set, cross-validation:

- **Uses all data for validation:** Every sample gets validated exactly once, maximizing data efficiency
- **Reduces variance:** Averages results across multiple splits, making performance estimates more stable and reliable
- **Avoids lucky/unlucky splits:** A single holdout might coincidentally be easy or hard to predict; cross-validation averages out this randomness

The tradeoff is computational cost—cross-validation requires training multiple models instead of one.
