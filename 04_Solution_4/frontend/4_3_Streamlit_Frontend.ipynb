# 4.3 Minimal Streamlit Frontend

The Streamlit application collects four measurements, sends them to the FastAPI backend as JSON, and displays the returned prediction.

## Folder structure

```text
frontend/
├── __init__.py
├── api_client.py
├── app.py
└── 4_3_Streamlit_Frontend.ipynb
```

`app.py` imports `request_prediction` from `api_client.py`. The UI therefore does not need to know the details of HTTP requests and error handling.

## Test the API client

First start the FastAPI backend in a separate terminal. The notebook then imports the same function used by the Streamlit application.

````python
import sys
from pathlib import Path

import requests

CURRENT_DIR = Path.cwd()
if (CURRENT_DIR / 'api_client.py').exists():
  FRONTEND_DIR = CURRENT_DIR
elif (CURRENT_DIR / 'frontend').exists():
  FRONTEND_DIR = CURRENT_DIR / 'frontend'
else:
  FRONTEND_DIR = CURRENT_DIR / '04_Solution_4' / 'frontend'

if str(FRONTEND_DIR) not in sys.path:
  sys.path.insert(0, str(FRONTEND_DIR))

from api_client import request_prediction
````

````python
payload = {
  'sepal_length_cm': 5.1,
  'sepal_width_cm': 3.5,
  'petal_length_cm': 1.4,
  'petal_width_cm': 0.2,
}

try:
  result = request_prediction(payload)
except requests.RequestException as error:
  print('Start the FastAPI backend with `python run.py` first.')
  print(error)
else:
  print(result)
````

## Start Streamlit

Open a second terminal in this frontend folder and run:

```text
streamlit run app.py
```

If the Conda environment does not expose the `streamlit` command on `PATH`, use:

```text
python -m streamlit run app.py
```

Streamlit normally opens `http://localhost:8501` in the browser.

## Run both services

```text
Terminal 1
cd backend
python run.py

Terminal 2
cd frontend
streamlit run app.py
```

The browser communicates with Streamlit on port 8501. The Streamlit Python process sends prediction requests to FastAPI on port 8000.
