# NumPy: numerical foundations for machine learning

NumPy provides efficient multidimensional arrays and vectorized numerical operations. Pandas, Matplotlib, scikit-learn, and many ML libraries build on these ideas.

The goal is not to memorize every function. Learn to reason about **shape**, **dtype**, **axis**, and **broadcasting**---the four words responsible for both elegant ML code and many mysterious error messages.

## 1. Arrays versus Python lists

```python
import numpy as np


python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print(python_list * 2)
print(numpy_array * 2)
print(type(numpy_array))
```

A list can mix arbitrary objects. A NumPy array usually stores one dtype, enabling compact storage and fast operations.

```python
integer_scores = np.array([67, 69, 99], dtype=np.int64)
float_scores = integer_scores.astype(np.float64)

print(integer_scores, integer_scores.dtype)
print(float_scores, float_scores.dtype)
```

## 2. Creating arrays and understanding shape

Important attributes are `ndim`, `shape`, `size`, and `dtype`. A two-dimensional array is commonly interpreted as rows by columns.

```python
zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
sequence = np.arange(0, 12, 2)
even_spacing = np.linspace(0, 1, 5)
score_matrix = np.array([
  [67, 69, 99],
  [42, 73, 91],
])

print(zeros)
print(ones)
print(sequence)
print(even_spacing)
print(score_matrix)
print("ndim:", score_matrix.ndim)
print("shape:", score_matrix.shape)
print("size:", score_matrix.size)
print("dtype:", score_matrix.dtype)
```

## 3. Indexing, slicing, and boolean masks

Each dimension receives an index or slice, separated by commas.

```python
scores = np.array([
  [67, 69, 99],
  [42, 73, 91],
  [55, 49, 84],
])

print(scores[0, 1])
print(scores[0])
print(scores[:, 1])
print(scores[:2, 1:])
print(scores[-1, -1])
```

```python
passing_mask = scores >= 50
print(passing_mask)
print(scores[passing_mask])

high_scores = scores[(scores >= 80) & (scores <= 90)]
print(high_scores)
```

### Views versus copies

Basic slices often return a **view** into the original array. Changing the view can change the source. Use `.copy()` when you need independence.

```python
original = np.array([67, 69, 99, 91])
view = original[1:3]
view[0] = 420
print("After changing view:", original)

original = np.array([67, 69, 99, 91])
independent = original[1:3].copy()
independent[0] = 420
print("After changing copy:", original)
```

## 4. Vectorization

Vectorized expressions operate on entire arrays without a Python loop. This is usually clearer and much faster for numerical work.

```python
scores = np.array([42.0, 67.0, 69.0, 99.0])
curved_scores = np.clip(scores * 1.05 + 2, 0, 100)
centered_scores = scores - scores.mean()

print(curved_scores)
print(centered_scores)
print(np.sqrt(scores))
print(np.round(scores / 7, 2))
```

## 5. Broadcasting

Broadcasting combines arrays with compatible shapes. Dimensions are compared from right to left. They are compatible when equal or when one of them is `1`.

```python
features = np.array([
  [6.0, 80.0, 8.0],
  [4.0, 67.0, 6.0],
  [9.0, 95.0, 10.0],
])
column_offsets = np.array([1.0, 10.0, 2.0])

adjusted = features - column_offsets
print(adjusted)
print("Shapes:", features.shape, column_offsets.shape, adjusted.shape)
```

A common ML operation standardizes each feature column. `keepdims=True` preserves a `(1, features)` shape and makes broadcasting visible.

```python
means = features.mean(axis=0, keepdims=True)
standard_deviations = features.std(axis=0, keepdims=True)
standardized = (features - means) / standard_deviations

print("Means:", means)
print("Standard deviations:", standard_deviations)
print("Standardized:", np.round(standardized, 2))
```

## 6. Aggregation and axes

With a 2D array, `axis=0` collapses rows and returns one result per column. `axis=1` collapses columns and returns one result per row.

```python
scores = np.array([
  [67, 69, 99],
  [42, 73, 91],
  [55, 49, 84],
])

print("Overall mean:", scores.mean())
print("Column means:", scores.mean(axis=0))
print("Row means:", scores.mean(axis=1))
print("Minimums:", scores.min(axis=0))
print("Maximums:", scores.max(axis=0))
print("Passing count:", np.count_nonzero(scores >= 50))
```

### Missing numerical values

`np.nan` represents a missing floating-point value. Ordinary aggregations propagate it; `nanmean`, `nansum`, and related functions ignore it.

```python
hours = np.array([4.2, 6.9, np.nan, 8.8])
print(np.mean(hours))
print(np.nanmean(hours))
print(np.isnan(hours))
```

## 7. Reshaping, transposing, and combining arrays

```python
sequence = np.arange(1, 13)
matrix = sequence.reshape(3, 4)
print(matrix)
print("Transposed:", matrix.T)
print("Flattened copy:", matrix.flatten())
print("Flattened view when possible:", matrix.ravel())
```

```python
left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])

print("Vertical stack:", np.vstack([left, right]))
print("Horizontal stack:", np.hstack([left, right]))
print("Concatenate rows:", np.concatenate([left, right], axis=0))
print("Concatenate columns:", np.concatenate([left, right], axis=1))
```

## 9. Linear algebra for ML

A dot product multiplies matching elements and sums the results. Matrix multiplication applies many dot products at once. In a simple linear model, predictions are `X @ weights + bias`.

```python
student_features = np.array([6.0, 80.0, 8.0])
weights = np.array([1.5, 0.2, 3.0])
bias = 5.0

prediction = student_features @ weights + bias
print("Predicted score:", prediction)

feature_matrix = np.array([
  [6.0, 80.0, 8.0],
  [4.0, 67.0, 6.0],
  [9.0, 95.0, 10.0],
])
predictions = feature_matrix @ weights + bias
print(predictions)
```

```python
identity = np.eye(3)
print(feature_matrix @ identity)
print("Feature covariance:", np.cov(feature_matrix, rowvar=False))
```

## 10. Pandas interoperability

DataFrames use NumPy arrays internally for many operations. Convert selected numeric columns with `.to_numpy()` and rebuild a DataFrame only when labels are needed again.

```python
from pathlib import Path
import pandas as pd


candidate_paths = [
    parent / "data" / "aai_student_data.csv"
    for parent in (Path.cwd(), *Path.cwd().parents)
]
data_path = next((path for path in candidate_paths if path.exists()), None)
if data_path is None:
  raise FileNotFoundError("Could not find the synthetic AAI dataset.")

students = pd.read_csv(data_path)
numeric_columns = ["attendance_percent", "assignments_completed", "exam_score"]
numeric_array = students[numeric_columns].to_numpy(dtype=float)

print(numeric_array.shape)
print(np.round(numeric_array.mean(axis=0), 2))

round_trip = pd.DataFrame(numeric_array[:3], columns=numeric_columns)
round_trip
```

## Takeaway

Before every NumPy operation, ask:

1. What is the array's shape?
2. What is its dtype?
3. Which axis should be reduced?
4. Will broadcasting do what I intend?
5. Am I working with a view or an independent copy?
