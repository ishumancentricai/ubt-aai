## 3. Collections

Collections store multiple values. Choosing the right collection makes code clearer:

| Type | Ordered | Mutable | Duplicates | Typical use |
|---|---:|---:|---:|---|
| `list` | yes | yes | yes | a sequence that changes |
| `tuple` | yes | no | yes | a fixed record |
| `set` | no guaranteed display order | yes | no | unique values and set logic |
| `dict` | insertion-ordered | yes | keys are unique | named key-value data |

### 3.1 Lists

### 3.3 Sets and frozen sets

Sets remove duplicates and support union (`|`), intersection (`&`), difference (`-`), and symmetric difference (`^`). A `frozenset` is immutable.

````python
group_a = {"Python", "Pandas", "Coffee"}
group_b = {"Python", "NumPy", "Coffee"}

print(group_a | group_b)
print(group_a & group_b)
print(group_a - group_b)
print(group_a ^ group_b)

group_a.add("Matplotlib")
group_a.discard("Nonexistent topic") # No error if missing.
fixed_topics = frozenset({"Python", "Git"})
print(group_a, fixed_topics)
````

## 4. Operators and decisions

Arithmetic operators include `+`, `-`, `*`, `/`, `//`, `%`, and `**`. Comparisons produce booleans. `and`, `or`, and `not` combine conditions. `in` checks membership. `is` checks object identity and is especially useful with `None`.

````python
a = 69
b = 4
print(a + b, a - b, a * b, a / b)
print(a // b, a % b, a ** 2)
print(a > b, a == 69, a != b)

has_enough_points = a >= 50
submitted_project = True
print(has_enough_points and submitted_project)
print("AI" in "Applied Artificial Intelligence")
print(None is None)
````

## 5. Loops

A `for` loop visits items in an iterable. `enumerate()` supplies an index. `zip()` combines iterables. A `while` loop repeats while a condition remains true.

````python
topics = ["Python", "Pandas", "NumPy"]

for position, topic in enumerate(topics, start=1):
  print(position, topic)

scores = [67, 69, 99]
for topic, score in zip(topics, scores):
  print(f"{topic}: {score}")
````

````python
countdown = 3
while countdown > 0:
  print(countdown)
  countdown -= 1
else:
  print("Live demo time!")
````

`break` exits a loop, `continue` skips to the next iteration, and `pass` is an explicit do-nothing placeholder.

````python
for number in range(1, 10):
  if number == 4:
    continue
  if number == 7:
    break
  print(number)

if True:
  pass # Useful while sketching a block, but remove it when real logic exists.
````

## 7. Functions

Functions package reusable behavior. Parameters are inputs. `return` sends a result back. A function that reaches the end without `return` returns `None`.

````python
def final_score(exam: float, project: float, exam_weight: float = 0.6) -> float:
  '''Calculate a weighted score between exam and project.'''
  project_weight = 1 - exam_weight
  return exam * exam_weight + project * project_weight


result = final_score(69, 99)
print(f"Final score: {result:.1f}")
````

Arguments may be positional or named. Use `*args` for additional positional arguments and `**kwargs` for additional named arguments. These tools are useful, but explicit parameters are easier to understand when the interface is known.

````python
def average(*values: float) -> float:
  return sum(values) / len(values)


def describe_student(name: str, **details: object) -> str:
  detail_text = ", ".join(f"{key}={value}" for key, value in details.items())
  return f"{name}: {detail_text}"


print(average(67, 69, 99))
print(describe_student("Ada", program="Mathematics", caffeinated=True))
````

Variables created inside a function are local. Reading globals is possible, but passing inputs explicitly usually makes code easier to test.

````python
course_code = "AAI"

def create_label(number: int) -> str:
  local_label = f"{course_code}-{number:03d}"
  return local_label


print(create_label(69))
````

## 8. Errors, exceptions, and assertions

Exceptions report problems during execution. Catch only errors you can handle meaningfully, and prefer specific exception types.

````python
raw_score = "sixty-nine"

try:
  numeric_score = float(raw_score)
except ValueError as error:
  print(f"Could not convert the score: {error}")
  numeric_score = None
else:
  print("Conversion worked.")
finally:
  print("Conversion attempt finished.")

print(numeric_score)
````

Assertions document internal assumptions during development. Do not use them as the only validation for untrusted user input.

````python
probability = 0.69
assert 0 <= probability <= 1, "Probability must be between 0 and 1."
print("Probability is valid.")
````

## 11. Lambda functions, recursion, generators, and decorators

````python
records = [("Ada", 69), ("Grace", 99), ("Guido", 67)]
print(sorted(records, key=lambda record: record[1], reverse=True))

def factorial(number: int) -> int:
  if number <= 1:
    return 1
  return number * factorial(number - 1)


def countdown(start: int):
  while start >= 0:
    yield start
    start -= 1


print(factorial(5))
print(list(countdown(3)))
````

````python
def announce_call(function):
  def wrapper(*args, **kwargs):
    print(f"Calling {function.__name__}")
    return function(*args, **kwargs)
  return wrapper


@announce_call
def lecturer_confidence() -> int:
  return 420


print(lecturer_confidence())
````
